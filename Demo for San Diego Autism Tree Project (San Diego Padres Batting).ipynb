```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

**Read in the dataset**

```python
padres.isnull().sum()
```

```python
duplicated = padres[padres.duplicated()]
```

```python
duplicated
```

```python
padres["Left_Handed_Hitter"] = np.where(padres["Dominant_Hand"] == "Left", 1, 0).astype(bool)
```

```python
padres["Right_Handed_Hitter"] = np.where(padres["Dominant_Hand"] == "Right", 1, 0).astype(bool)
```

```python
padres["Switch_Hitter"] = padres["Switch_Hitter"].astype(bool)
```

```python
padres
```

**Drop Records for Position equals Pitcher, most pitchers stink and cannot hit**

```python
Index = padres[(padres["Position"] == "P")].index
padres.drop(Index, inplace=True)
```

```python
padres
```

```python
padres["Position"].value_counts()
```

**Create Column for all infielders in padres history**

```python
padres_infielder = padres[(padres["Position"] == "C") | (padres["Position"] == "1B") | (padres["Position"] == "2B") | (padres["Position"] == "SS") | (padres["Position"] == "3B") | (padres["Position"] == "IF") | (padres["Position"] == "CI") | (padres["Position"] == "MI")]
```

```python
padres["Infielder"] = padres["Position"].isin(padres_infielder["Position"]).astype(bool)
```

```python
padres
```

**Create column for all outfielders in padres history**

```python
padres_outfielder = padres[(padres["Position"] == "OF") | (padres["Position"] == "LF") | (padres["Position"] == "RF")]
```

```python
padres["Outfielder"] = padres["Position"].isin(padres_outfielder["Position"]).astype(bool)
```

```python
padres
```

```python
padres_utility = padres[padres["Position"] == "UT"]
```

```python
padres["Utility"] = padres["Position"].isin(padres_utility["Position"]).astype(bool)
```

```python
padres
```

```python
padres_dh = padres[padres["Position"] == "DH"]
```

```python
padres["DH"] = padres["Position"].isin(padres_dh["Position"]).astype(bool)
```

```python
padres
```

```python
padres.dtypes
```

```python
padres["Total_Extra_Base_Hits"] = padres["Doubles"] + padres["Triples"] + padres["Home_Runs"]
```

```python
padres
```

**Classification Model**
X: Age	Games	Plate_Appearances	At_Bats	Runs	Hits	Doubles	Triples	Home_Runs	Runs_Batted_In	Stolen_Bases	Caught_Stealing	Base_On_Balls	Strikeouts	Batting_Average	On_Base_Percentage	Slugging_Percentage	On_Base_Plus_Slugging_Percentage	On_Base_Plus_Slugging_Percentage_Plus	Total_Bases	Double_Plays_Grounded_Into	Times_Hit_By_Pitch	Sacrifice_Hits	Sacrifice_Flies	Intentional_Bases_on_Balls

- Base Logistic Regression model 72.02%
- grid search regression model: 73.87%
- K cross fold validation Logistic Regression Model 81.72%
- Logistic Regression Cross Fold Validation 74.10%

```python
numerics = padres[['Age', 'Games', 'Plate_Appearances', 'At_Bats', 'Runs', 'Hits', 'Doubles', 'Triples', 'Home_Runs', 'Runs_Batted_In', 'Stolen_Bases', 'Caught_Stealing', 'Base_On_Balls', 'Strikeouts', 'Batting_Average', 'On_Base_Percentage', 'Slugging_Percentage', 'On_Base_Plus_Slugging_Percentage', 'On_Base_Plus_Slugging_Percentage_Plus', 'Total_Bases', 'Double_Plays_Grounded_Into', 'Times_Hit_By_Pitch', 'Sacrifice_Hits', 'Sacrifice_Flies', 'Intentional_Bases_on_Balls', 'Left_Handed_Hitter']]
```

```python
X = numerics.iloc[:, :-1].values
```

```python
Y = numerics.iloc[:, -1].values
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.25, random_state=42, shuffle=True)
```

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
```

```python
from sklearn.linear_model import LogisticRegression
classifier = LogisticRegression()
classifier.fit(X_train, Y_train)
```

```python
y_pred = classifier.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix
cm = confusion_matrix(Y_test, y_pred)
print(cm)
accuracy_score(Y_test, y_pred)
```

```python
param_grid = {"C": [0.001, 0.001, 0.1, 1, 10, 100],
             'penalty': ['l1', 'l2'],
             'solver': ['liblinear', 'saga']}
```

```python
from sklearn.linear_model import LogisticRegression
classifier = LogisticRegression(max_iter=1000)
```

```python
from sklearn.model_selection import GridSearchCV
grid_search = GridSearchCV(estimator=classifier, param_grid=param_grid, scoring='accuracy', n_jobs=-1)
```

```python
grid_search.fit(X_train, Y_train)
```

```python
print("Best Parameters:", grid_search.best_params_)
print("Best Score:", grid_search.best_score_)
```

**Use Logistic Regression Cross Validation**

```python
from sklearn.linear_model import LogisticRegressionCV
```

```python
clf = LogisticRegressionCV(cv=10,
                          scoring='accuracy',
                          random_state=0,
                          n_jobs = -1,
                          verbose=3,
                          max_iter=1000).fit(X_train, Y_train)
```

```python
clf.score(X_train, Y_train)
```

**K-Fold Cross Validation**

```python
from sklearn.model_selection import KFold
n_splits = 10
cv = KFold(n_splits=n_splits)
```

```python
from sklearn.model_selection import cross_val_score
classifier1 = LogisticRegression()
scores=cross_val_score(
    classifier1, X_train, Y_train, scoring='accuracy', cv=cv, n_jobs=-1)
```
